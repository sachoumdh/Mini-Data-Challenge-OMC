# Mini Data Challenge

A secret message is hidden inside secret_data.csv. Your mission:

1- Sort by time_spent

2- Map each click to a letter

3- Piece together the hidden message

```python
import pandas as pd
```

## Solution

```python
# Load the CSV
df = pd.read_csv("secret_data.csv")
df.head()
```

```python
# Displays only the time_on_platform column
print(df['time_on_platform'])
```

## Sort by time_spent

```python
# Sort by time spent in ascending order
df_sorted = df.sort_values(by=['time_on_platform'], ascending=[True])

# Display the sorted table
print(df_sorted)
```

## Map each click to a letter 

```python
# Creation of a dictionary where each number represents a specific letter in alphabetical order
map_dictionary = {1: 'A', 2: 'B', 3: 'C', 4: 'D', 5: 'E', 6: 'F', 7: 'G', 8: 'H', 9: 'I', 10: 'J', 11: 'K', 12: 'L', 13: 'M', 14: 'N', 15: 'O', 16: 'P', 17: 'Q', 18: 'R', 19: 'S', 20: 'T', 21: 'U', 22: 'V', 23: 'W', 24: 'X', 25: 'Y', 26: 'Z'}

# Creation of a new column "letter"
df_sorted['letter'] = df_sorted['clicks'].map(map_dictionary)

# Display the new table with the column "letter" 
print(df_sorted)
```

## Piece together the hidden message

```python
# Creation of a list that contains the elements of the column "letter"
list_letter = df_sorted['letter'].tolist()

# Display the list (the hidden message is in the list)
print(list_letter)
```

```python
print(" ".join(list_letter))
```
